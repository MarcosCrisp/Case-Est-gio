## Análise das Regras de Negócios

```python
limite = 5_000_000

acima_sem_dupla = (df['Valor_R$'] > limite) & (df['Dupla_Aprovacao'] != 'Sim')
abaixo_com_dupla = (df['Valor_R$'] < limite) & (df['Dupla_Aprovacao'] == 'Sim')

regra1_atendida = (
    ((df['Valor_R$'] > limite) & (df['Dupla_Aprovacao'] == 'Sim'))
    | ((df['Valor_R$'] < limite) & (df['Dupla_Aprovacao'] == 'Não'))
)
df['Regra1'] = np.where(regra1_atendida, 'Sim', 'Não')

df['Regra1_Descricao'] = np.select(
    [acima_sem_dupla, abaixo_com_dupla],
    [
        'Valor acima de R$ 5 milhões sem dupla aprovação',
        'Valor abaixo de R$ 5 milhões com dupla aprovação'
    ],
    default='Sem divergência identificada'
)

df
```

```python
liquidada_com_data = (
    df['Status'].isin(['Liquidada', 'Concluída'])
    & df['Data_Liquidacao'].notna()
)
cancelada_ou_pendente_sem_data = (
    df['Status'].isin(['Cancelada', 'Pendente'])
    & df['Data_Liquidacao'].isna()
)

df['Verificacao_liquidada'] = (
    liquidada_com_data | cancelada_ou_pendente_sem_data
)

status = df['Status']
data_liquidacao = df['Data_Liquidacao']

df['Verificacao_Liquidada_Descricao'] = np.select(
    [
        df['Verificacao_liquidada'],
        (status == 'Liquidada') & data_liquidacao.isna(),
        (status == 'Concluída') & data_liquidacao.isna(),
        (status == 'Cancelada') & data_liquidacao.notna(),
        (status == 'Pendente') & data_liquidacao.notna()
    ],
    [
        'Normal',
        'Liquidada sem data',
        'Concluída sem data',
        'Cancelada com data',
        'Pendente com data'
    ],
    default='Status não previsto'
)

df[['Status', 'Data_Liquidacao', 'Verificacao_liquidada', 'Verificacao_Liquidada_Descricao']]
```

## Operações Repetidas

```python
# Analisando quais são os ID's acima de 1
x = df.groupby('ID_Operacao').size().reset_index(name='counts')
x = x[x['counts'] > 1]
x
```

```python
df_id_operacao_repetidos = df[df['ID_Operacao'].isin(x['ID_Operacao'])]
df_id_operacao_repetidos.sort_values(by=['ID_Operacao', 'Data_Operacao'], inplace=True)
df_id_operacao_repetidos
```

```python
df_intervalos = df_id_operacao_repetidos[['ID_Operacao', 'Data_Operacao']].copy()

df_intervalos = df_intervalos.sort_values(
    ['ID_Operacao', 'Data_Operacao']
)
df_intervalos['Data_Operacao_Anterior'] = df_intervalos.groupby('ID_Operacao')['Data_Operacao'].shift(1)
df_intervalos['Distancia_dias'] = (
    df_intervalos['Data_Operacao'] - df_intervalos['Data_Operacao_Anterior']
).dt.days

df_distancia_dias = df_intervalos.dropna(subset=['Data_Operacao_Anterior']).copy()

print('Intervalos entre ocorrências consecutivas:')
display(df_distancia_dias)
```
